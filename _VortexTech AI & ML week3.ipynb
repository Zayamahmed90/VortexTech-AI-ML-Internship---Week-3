```python
# ============================================
# WEEK 3 - REGRESSION AND CLUSTERING
# VortexTech AI & ML Internship
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

import warnings
warnings.filterwarnings("ignore")
```

```python
# Load the dataset
df = pd.read_csv("train.csv")

# Display first 5 rows
df.head()
```

```python
# Dataset shape
print("Dataset Shape:", df.shape)

# Column names
print("\nColumns:")
print(df.columns.tolist())

# Dataset information
print("\nDataset Information:")
df.info()
```

```python
# Count missing values
missing_values = df.isnull().sum()

# Show only columns having missing values
missing_values = missing_values[missing_values > 0]

print("Columns with missing values:")
print(missing_values.sort_values(ascending=False))
```

```python
# Separate target variable
X = df.drop("SalePrice", axis=1)
y = df["SalePrice"]

# Identify numerical and categorical columns
numeric_columns = X.select_dtypes(include=["int64", "float64"]).columns
categorical_columns = X.select_dtypes(include=["object"]).columns

# Fill missing numerical values with median
X[numeric_columns] = X[numeric_columns].fillna(
    X[numeric_columns].median()
)

# Fill missing categorical values with mode
for column in categorical_columns:
    X[column] = X[column].fillna(X[column].mode()[0])

# Convert categorical columns into dummy variables
X = pd.get_dummies(X, drop_first=True)

print("Shape after preprocessing:", X.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# Create the regression model
model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

# Train the model
model.fit(X_train, y_train)

print("Random Forest Regressor trained successfully!")
```

```python
# Predict house prices
y_pred = model.predict(X_test)

# Display some predictions
prediction_comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred
})

prediction_comparison.head(10)
```

```python
# Calculate Mean Squared Error
mse = mean_squared_error(y_test, y_pred)

# Calculate RMSE
rmse = np.sqrt(mse)

# Calculate R2 score
r2 = r2_score(y_test, y_pred)

print("Regression Results")
print("--------------------------")
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
# Select numeric features for clustering
cluster_data = df[
    ["GrLivArea", "OverallQual", "GarageCars"]
].copy()

# Check missing values
print(cluster_data.isnull().sum())

# Fill missing values
cluster_data = cluster_data.fillna(
    cluster_data.median()
)

cluster_data.head()
```

```python
# Create scaler
scaler = StandardScaler()

# Scale clustering features
X_scaled = scaler.fit_transform(cluster_data)

print("Data scaled successfully!")
```

```python
# Store inertia values
inertia = []

# Test cluster numbers from 1 to 10
for k in range(1, 11):
    
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    
    inertia.append(kmeans.inertia_)

# Display inertia values
for k, value in zip(range(1, 11), inertia):
    print(f"K = {k}, Inertia = {value:.2f}")
```
